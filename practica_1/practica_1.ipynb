## Introducción

En el desarrollo de proyectos de robótica es necesario utilizar diferentes herramientas de software para **crear archivos, automatizar tareas, documentar procedimientos y compartir el trabajo**. Aunque estas herramientas tienen funciones diferentes, su combinación permite establecer flujos de trabajo reproducibles y facilita la colaboración.

- <img src="imagenes/Ubuntu-logo.png" height=60/> **Ubuntu** es un sistema operativo basado en Linux ampliamente utilizado en robótica. El manejo de la terminal permite ejecutar programas, administrar archivos y automatizar tareas sin depender exclusivamente de interfaces gráficas.

- **Bash (Bourne Again SHell)** es un intérprete de comandos utilizado habitualmente en sistemas Linux. Además de permitir ejecutar comandos de manera interactiva, permite crear **scripts de Bash**, es decir, archivos de texto que contienen una secuencia de comandos que puede ejecutarse de manera automatizada.

- <img src="imagenes/Jupyter-logo.png" height=60/> **Jupyter** proporciona un entorno interactivo basado en notebooks. Un notebook está formado por celdas que pueden contener texto, ecuaciones, código y resultados. En esta práctica se utilizará para documentar el trabajo y ejecutar comandos de Bash desde el mismo documento.

- **Markdown** es un lenguaje de marcado ligero que permite dar formato al texto mediante una sintaxis sencilla. Jupyter utiliza Markdown para crear títulos, listas, tablas, enlaces y ecuaciones dentro de las celdas de texto.

- <img src="imagenes/Git-logo.png" height=60/> **Git** es un sistema de control de versiones que permite registrar cambios en los archivos de un proyecto, recuperar versiones anteriores y trabajar mediante ramas.

- <img src="imagenes/Github-logo.png" height=60/> **GitHub** es una plataforma que permite alojar repositorios Git en línea y facilita el trabajo colaborativo mediante repositorios, ramas, colaboradores y herramientas de revisión.

En robótica, estas herramientas pueden utilizarse conjuntamente. Por ejemplo, un alumno puede crear un script de Bash para preparar una carpeta de trabajo, documentar el procedimiento en Jupyter y finalmente almacenar el notebook y los scripts en un repositorio de GitHub.

### Responder de forma breve las siguientes preguntas:

- En Linux, ¿qué es una **shell** y qué relación existe entre `sh` y `bash`?
> Shell: Es un programa que esta entre el usuario y el kernel. Recibe comandos y pide que se ejecuten.
> 'sh' fue uno de los primeros shell que existieron, bash es una versión actual y mejorada, incluye herramientas como el historial de commandos y autocompletado con TAB

- ¿Qué es un **script de Bash** y cuál es la diferencia entre ejecutar un comando directamente en la terminal y ejecutarlo mediante un script?
> Un Script de Bash es un archivo que tiene una secuencia de comandos de Linux y se leen y ejecutan de forma automática. Al ejecutarlo en la terminal es interactivo, nosotros decidimos que comandos ejecutar y se hace uno por uno. Al ejecutar en script se hace de manera automática y todos de forma seguida.

- ¿Qué es **Jupyter** y qué tipo de información puede contener una celda de un notebook?
> Entorno de trabajo basado en Notebooks, el notebook esta conformador por celdas. Estas celdas pueden tener código, resultado de código de otra celda, o texto formateado usando Markdowns.

- ¿Qué es **Markdown** y para qué se utiliza dentro de Jupyter?
> Es una forma de dar formato a textos dentro de Notebooks de Jupyter por medio de símbolos sencillos que le dan formato al texto, sin usar una síntaxis complicada.

- ¿Cuál es la diferencia entre **Git** y **GitHub**?
> Git es el software de control de verisiones que se encuentra de forma local en la computadora. GitHub es una plataforma que guarda repositorios (proyectos) que usan Git en internet.

### 1. Markdown y Ecuaciones

- **Forma general de la matriz de transformación homogénea de un sistema $j$ respecto a un sistema $i$:**
$$ T_j^i = \begin{bmatrix} R_j^i & p_j^i \\ \mathbf{0} & 1 \end{bmatrix} $$

- **Forma general de la matriz de rotación de un sistema $j$ respecto a un sistema $i$ respecto al eje $z$:**
$$ R_j^i(z, \theta) = \begin{bmatrix} \cos\theta & -\sin\theta & 0 \\ \sin\theta & \cos\theta & 0 \\ 0 & 0 & 1 \end{bmatrix} $$

- **Forma general del vector de posición de un sistema $j$ respecto a un sistema $i$:**
$$ p_j^i = \begin{bmatrix} p_x \\ p_y \\ p_z \end{bmatrix} $$

- **Forma general del vector de postura para un manipulador contenido en el plano $XY$:**
$$ \mathbf{x} = \begin{bmatrix} x \\ y \\ \phi \end{bmatrix} $$

- **Forma general del vector de juntas para un manipulador RRR:**
$$ \mathbf{q} = \begin{bmatrix} q_1 \\ q_2 \\ q_3 \end{bmatrix} = \begin{bmatrix} \theta_1 \\ \theta_2 \\ \theta_3 \end{bmatrix} $$


### 2. Manejo de Ubuntu y Bash

En esta primera parte se creará un **script de Bash** que automatice la creación y eliminación de una estructura de carpetas.

El script deberá:

- Desplegar un mensaje de inicio.

- Eliminar la carpeta `Practica1`, incluyendo sus subcarpetas y archivos.
- Dar permisos de ejecución al script y ejecutarlo.

**Importante:** antes de utilizar comandos que eliminen archivos, comprobar que la ruta corresponde únicamente a la carpeta creada para esta práctica.

### Código del script de Bash (`practica1.sh`)

```bash
#!/bin/bash

# 1. Desplegar mensaje de inicio
echo "Iniciando configuración de directorios para la Práctica 1..."

# 2. Moverse a la carpeta de usuario
cd $HOME
echo "Directorio actual: $(pwd)"

# 3 y 4. Verificar si existe la carpeta Practica1 y crearla
if [ -d "Practica1" ]; then
    echo "La carpeta 'Practica1' ya existe. Eliminándola..."
    rm -rf Practica1
fi

mkdir Practica1
echo "Carpeta 'Practica1' creada exitosamente."

# 5. Crear carpeta Letras y sus archivos
mkdir Practica1/Letras
touch Practica1/Letras/a.txt Practica1/Letras/b.txt Practica1/Letras/c.txt

# 6. Crear carpeta Integrantes y sus archivos con los nombres del equipo
mkdir Practica1/Integrantes
touch Practica1/Integrantes/AndresIbarraVazquez.txt
touch Practica1/Integrantes/JoseAugustoArenasHernandez.txt
touch Practica1/Integrantes/AxelAlejandroMartinezRamirez.txt

# 7. Mostrar un diagrama de la estructura
echo "--------------------------------------"
echo "Estructura generada:"
if command -v tree &> /dev/null; then
    tree Practica1
else
    echo "(El comando 'tree' no está instalado. Usando 'find' como alternativa visual)"
    find Practica1 | sed -e "s/[^-][^\/]*\//  |/g" -e "s/|\([^ ]\)/|-\1/"
fi
echo "--------------------------------------"

# 8. Eliminar la carpeta Practica1 y su contenido
echo "Eliminando la estructura para mantener limpio el entorno..."
rm -rf Practica1

echo "¡Script finalizado con éxito!"


### Explicación de los bloques principales del script

- **Cabecera (`#!/bin/bash`):** Indica al sistema operativo que debe utilizar el intérprete Bash para leer y ejecutar los comandos de este archivo.
- **Navegación e Inicio (`cd $HOME`, `echo`):** Imprime mensajes informativos en la terminal y asegura que la estructura se cree siempre dentro del directorio principal del usuario, sin importar desde dónde se ejecutó el script.
- **Condicional Inicial (`if [ -d ... ]`):** Verifica si la carpeta `Practica1` ya existe de una ejecución anterior. Si existe, la borra para evitar duplicados o errores y permitir que el script corra en limpio.
- **Creación de Estructura (`mkdir`, `touch`):** Se encarga de construir la jerarquía de directorios solicitada. `mkdir` crea la carpeta principal y las subcarpetas, mientras que `touch` genera todos los archivos de texto (letras e integrantes del equipo) vacíos simultáneamente.
- **Visualización (`tree` / `find`):** Despliega el árbol de directorios para comprobar gráficamente que los archivos están en el lugar correcto. Incluye una estructura lógica (`if`) que usa un sustituto visual con `find` en caso de que el comando `tree` no esté instalado en la máquina.
- **Limpieza (`rm -rf`):** Elimina de forma recursiva (`-r`) y forzada (`-f`) la carpeta `Practica1` y todo su contenido, cumpliendo con la instrucción de no dejar rastros en el sistema al terminar.


### Captura de la terminal:

![Captura de la ejecución del script](imagenes/CapturaP1.png)

### 3. Git y GitHub

Generar un repositorio en línea llamado `Robotica-2027-1`, donde se subirán los archivos de la práctica. Agregar a los miembros del equipo como colaboradores del repositorio.

En la configuración del repositorio incluir un archivo `README.md`, y en él colocar:

- Una descripción del repositorio.
- Nombres de los integrantes del equipo.
- Una breve descripción de la organización de las prácticas.

Se recomienda primero crear el repositorio en línea, clonarlo en el ordenador y después incluir los archivos de la práctica.

Para utilizar el repositorio para todas las prácticas posteriores, se recomienda usar la siguiente estructura:

```text
Robotica-2027-1/
├── practica_1/
│   ├── imagenes/
│   │   └── p1_1.png
│   ├── scripts/
│   │   └── practica1.sh
│   └── practica_1.ipynb
├── README.md
└── .gitignore
```
